```python
import librosa
import noisereduce as nr
import soundfile as sf

# Load audio
audio, sr = librosa.load("Exact_File_Name_what_uploaded.wav", sr=None, mono=True)

print("Sample rate:", sr)
print("Duration:", len(audio) / sr, "seconds")

# Take a short section assumed to contain mostly background noise
noise_sample = audio[:int(0.5 * sr)]

# Noise reduction
clean_audio = nr.reduce_noise(
    y=audio,
    sr=sr,
    y_noise=noise_sample,
    stationary=True,
    prop_decrease=1.0
)

# Save
sf.write("Exact_File_Name_what_uploaded_clean.wav", clean_audio, sr)

print("✅ Created Exact_File_Name_what_uploaded_clean.wav")
```

```python
from IPython.display import Audio, display

print("ORIGINAL")
display(Audio("Exact_File_Name_what_uploaded.wav"))

print("CLEANED")
display(Audio("Exact_File_Name_what_uploaded_clean.wav"))
```
